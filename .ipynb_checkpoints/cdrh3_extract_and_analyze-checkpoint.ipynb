```python
import pandas as pd
import matplotlib.pyplot as plt
import logomaker
from Bio import SeqIO
import re

# === CONFIGURATION ===
TRAIN_FILE = "/home1/ZZHANG201@kgi.edu/GAN/zzGAN/gan/data/zz/train.fasta"   # Replace with your real file
GEN_FILE = "/project/animesh_ray_1465/Zihao/GAN/logs/trial4/20251203-153206/candidates_best_7k.fasta" # Replace with your real file
REGEX_PATTERN = r"C[A-Z]{3,30}WG"       # Your CDR3 regex

def extract_cdr3s(fasta_file, pattern):
    loops = []
    for record in SeqIO.parse(fasta_file, "fasta"):
        seq = str(record.seq).upper()
        match = re.search(pattern, seq)
        if match:
            loops.append(match.group(0))
    return loops

# 1. Load Data
train_loops = extract_cdr3s(TRAIN_FILE, REGEX_PATTERN)
gen_loops = extract_cdr3s(GEN_FILE, REGEX_PATTERN)

print(f"Found {len(train_loops)} training loops and {len(gen_loops)} generated loops.")

# 2. Pre-process for Logo (Align to Center)
# Logos need fixed length. We will align by the 'C' anchor and pad the end.
def pad_sequences(sequences, length=15):
    padded = []
    for s in sequences:
        if len(s) > length: s = s[:length] # Truncate long ones
        else: s = s + '-' * (length - len(s)) # Pad short ones
        padded.append(s)
    return padded

max_len = 18
train_padded = pad_sequences(train_loops, max_len)
gen_padded = pad_sequences(gen_loops, max_len)

# 3. Create Logos
fig, axes = plt.subplots(2, 1, figsize=(10, 6))

# Train Logo
train_counts = logomaker.alignment_to_matrix(train_padded)
logo_train = logomaker.Logo(train_counts, ax=axes[0], color_scheme='chemistry')
axes[0].set_title("Training Data: CDRH3 Sequence Grammar", fontsize=14)
axes[0].set_ylabel("Bits")

# Gen Logo
gen_counts = logomaker.alignment_to_matrix(gen_padded)
logo_gen = logomaker.Logo(gen_counts, ax=axes[1], color_scheme='chemistry')
axes[1].set_title("GAN Generated: CDRH3 Sequence Grammar", fontsize=14)
axes[1].set_ylabel("Bits")

plt.tight_layout()
plt.savefig("/home1/ZZHANG201@kgi.edu/GAN/viz/gan/Figure_B_SequenceLogos_new.png", dpi=300)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from Bio import SeqIO
from Bio.SeqUtils.ProtParam import ProteinAnalysis
import re

# === CONFIGURATION ===
TRAIN_FILE = "/home1/ZZHANG201@kgi.edu/GAN/zzGAN/gan/data/zz/train.fasta"   # Replace with your real file
GEN_FILE = "/project/animesh_ray_1465/Zihao/GAN/logs/trial4/20251203-153206/candidates_best_7k.fasta" # Replace with your real file
REGEX_PATTERN = r"C[^CW]{3,30}W"       # Your CDR3 regex

def get_properties(fasta_file, label, pattern):
    data = []
    for record in SeqIO.parse(fasta_file, "fasta"):
        seq = str(record.seq).upper()
        match = re.search(pattern, seq)
        if match:
            loop = match.group(0)
            # Calculate Properties using Biopython
            analysis = ProteinAnalysis(loop)
            
            # 1. Length
            length = len(loop)
            
            # 2. Hydrophobicity (Gravy)
            gravy = analysis.gravy()
            
            # 3. Charge (Isoelectric Point approx)
            charge = analysis.charge_at_pH(7.4)
            
            data.append({"Sequence": loop, "Length": length, "Hydrophobicity": gravy, "Charge": charge, "Source": label})
    return pd.DataFrame(data)

# 1. Load Data
df_train = get_properties(TRAIN_FILE, "Natural (Train)", REGEX_PATTERN)
df_gen = get_properties(GEN_FILE, "GAN (Generated)", REGEX_PATTERN)

# Combine
df_all = pd.concat([df_train, df_gen])

# 2. Plotting
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
sns.set_style("whitegrid")
palette = {"Natural (Train)": "gray", "GAN (Generated)": "orange"}

# Plot A: Length
sns.histplot(data=df_all, x="Length", hue="Source", kde=True, ax=axes[0], palette=palette, bins=15, alpha=0.6)
axes[0].set_title("CDRH3 Length Distribution")

# Plot B: Hydrophobicity
sns.histplot(data=df_all, x="Hydrophobicity", hue="Source", kde=True, ax=axes[1], palette=palette, bins=30, alpha=0.6)
axes[1].set_title("Hydrophobicity (GRAVY)")

# Plot C: Charge
sns.histplot(data=df_all, x="Charge", hue="Source", kde=True, ax=axes[2], palette=palette, bins=30, alpha=0.6)
axes[2].set_title("Net Charge (pH 7.4)")

plt.tight_layout()
plt.savefig("/home1/ZZHANG201@kgi.edu/GAN/viz/gan/Figure_C_Properties_new_regex.png", dpi=300)
plt.show()
```
