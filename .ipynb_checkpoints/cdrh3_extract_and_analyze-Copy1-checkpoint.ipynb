```python
import re
from collections import Counter

import pandas as pd
from Bio import SeqIO
from Bio.SeqUtils.ProtParam import ProteinAnalysis


# =========================================================
# CONFIG
# =========================================================
# These are NOT claimed as exact biological boundaries.
# They are rescue motifs for defining a graftable H3-like interval
# when canonical framework annotation has failed.
DEFAULT_RESCUE_MOTIFS = [
    "VTITAS",
    "VTIRAS",
    "VTISAS",
    "VRITAS",
]

# You can relax to 5/6 match later if needed:
# min_similarity = 0.833333
DEFAULT_MIN_SIMILARITY = 1.0


import re
from collections import Counter, defaultdict

import pandas as pd
from Bio import SeqIO


def _hamming_similarity(a, b):
    if len(a) != len(b):
        return 0.0
    return sum(x == y for x, y in zip(a, b)) / len(a)


def _find_terminal_right_w(seq, right_anchor_max_from_end=10):
    w_positions = [m.start() for m in re.finditer("W", seq)]
    terminal_ws = [p for p in w_positions if (len(seq) - 1 - p) <= right_anchor_max_from_end]
    return max(terminal_ws) if terminal_ws else None


def discover_rescue_motif_windows(
    file_path,
    k=6,
    right_anchor_max_from_end=10,
    search_dist_range=(8, 25),
):
    """
    Collect all k-mer windows upstream of the terminal W anchor, along with their distance to W.
    This is the raw scan before collapsing into motif families.
    """
    rows = []

    for record in SeqIO.parse(file_path, "fasta"):
        seq = str(record.seq).upper()
        right_w = _find_terminal_right_w(seq, right_anchor_max_from_end=right_anchor_max_from_end)
        if right_w is None:
            continue

        for dist in range(search_dist_range[0], search_dist_range[1] + 1):
            start = right_w - dist
            if start < 0 or start + k > len(seq):
                continue

            motif = seq[start:start + k]
            rows.append({
                "SequenceID": record.id,
                "Motif": motif,
                "DistanceToW": dist,
                "Start": start,
                "End": start + k,
            })

    return pd.DataFrame(rows)


def summarize_rescue_motif_windows(window_df):
    """
    Summarize raw motif windows by (Motif, DistanceToW).
    """
    if window_df.empty:
        return pd.DataFrame(columns=["Motif", "DistanceToW", "Count", "Percent"])

    total = len(window_df)
    out = (
        window_df.groupby(["Motif", "DistanceToW"])
        .size()
        .reset_index(name="Count")
        .sort_values(["Count", "Motif", "DistanceToW"], ascending=[False, True, True])
        .reset_index(drop=True)
    )
    out["Percent"] = out["Count"] / total * 100
    return out


def _are_shift_neighbors(m1, m2, min_overlap=5):
    """
    Returns True if two same-length motifs look like adjacent sliding windows
    from the same longer conserved block.
    For k=6, overlap of 5 is the typical one-step shift relation.
    """
    if len(m1) != len(m2):
        return False

    k = len(m1)

    # right shift: ABCDEF vs BCDEFG
    if m1[1:] == m2[:-1] and (k - 1) >= min_overlap:
        return True

    # left shift: BCDEFG vs ABCDEF
    if m1[:-1] == m2[1:] and (k - 1) >= min_overlap:
        return True

    return False


def _build_shift_graph(motifs):
    """
    Build connected components among motifs that are sliding-window neighbors.
    """
    motifs = sorted(set(motifs))
    adj = defaultdict(set)

    for i, m1 in enumerate(motifs):
        for j in range(i + 1, len(motifs)):
            m2 = motifs[j]
            if _are_shift_neighbors(m1, m2):
                adj[m1].add(m2)
                adj[m2].add(m1)

    visited = set()
    components = []

    for m in motifs:
        if m in visited:
            continue
        stack = [m]
        comp = []
        visited.add(m)

        while stack:
            cur = stack.pop()
            comp.append(cur)
            for nxt in adj[cur]:
                if nxt not in visited:
                    visited.add(nxt)
                    stack.append(nxt)

        components.append(sorted(comp))

    return components


def _merge_sliding_windows(component):
    """
    Merge a connected sliding-window component into one consensus block string.
    Assumes each motif differs by one-position shifts.
    """
    if len(component) == 1:
        return component[0]

    # Find leftmost motif as one that is not suffix-aligned to another
    motifs = list(component)

    left_candidates = []
    for m in motifs:
        is_left = True
        for other in motifs:
            if m == other:
                continue
            if other[1:] == m[:-1]:
                is_left = False
                break
        if is_left:
            left_candidates.append(m)

    start = sorted(left_candidates)[0] if left_candidates else sorted(motifs)[0]

    ordered = [start]
    used = {start}

    extended = True
    while extended:
        extended = False
        last = ordered[-1]
        for m in motifs:
            if m in used:
                continue
            if last[1:] == m[:-1]:
                ordered.append(m)
                used.add(m)
                extended = True
                break

    block = ordered[0]
    for m in ordered[1:]:
        block += m[-1]

    return block


def collapse_rescue_motif_families(
    window_df,
    anchor_center_dist=16,
    preferred_k=6,
):
    """
    Collapse overlapping/sliding-window motifs into motif families.
    For each family, recommend one anchor motif phase for extraction.
    """
    if window_df.empty:
        return pd.DataFrame(), pd.DataFrame()

    motif_counts = window_df["Motif"].value_counts().to_dict()
    motifs = list(motif_counts.keys())

    components = _build_shift_graph(motifs)

    family_rows = []
    motif_to_family = {}

    for fam_id, comp in enumerate(components, start=1):
        block = _merge_sliding_windows(comp)

        # all rows belonging to motifs in this family
        fam_df = window_df[window_df["Motif"].isin(comp)].copy()

        # summarize motif+dist inside family
        fam_md = (
            fam_df.groupby(["Motif", "DistanceToW"])
            .size()
            .reset_index(name="Count")
            .sort_values(["Count", "Motif", "DistanceToW"], ascending=[False, True, True])
            .reset_index(drop=True)
        )

        # choose recommended anchor motif:
        # prioritize motifs with length=preferred_k and distances near anchor_center_dist
        fam_md["DistPenalty"] = (fam_md["DistanceToW"] - anchor_center_dist).abs()

        best = fam_md.sort_values(
            ["Count", "DistPenalty", "Motif"],
            ascending=[False, True, True]
        ).iloc[0]

        recommended_anchor = best["Motif"]
        recommended_dist = int(best["DistanceToW"])

        total_family_hits = len(fam_df)
        unique_sequences = fam_df["SequenceID"].nunique()

        family_rows.append({
            "FamilyID": fam_id,
            "FamilyBlock": block,
            "MotifsInFamily": ", ".join(sorted(comp)),
            "NumMotifs": len(comp),
            "TotalHits": total_family_hits,
            "UniqueSeqsWithFamily": unique_sequences,
            "RecommendedAnchorMotif": recommended_anchor,
            "RecommendedDistanceToW": recommended_dist,
        })

        for m in comp:
            motif_to_family[m] = fam_id

    family_df = pd.DataFrame(family_rows).sort_values(
        ["TotalHits", "FamilyID"], ascending=[False, True]
    ).reset_index(drop=True)

    detailed_df = window_df.copy()
    detailed_df["FamilyID"] = detailed_df["Motif"].map(motif_to_family)

    return family_df, detailed_df


def recommend_default_rescue_motifs(
    family_df,
    top_n_families=3,
):
    """
    Return a compact DEFAULT_RESCUE_MOTIFS list using one recommended anchor motif per family.
    """
    if family_df.empty:
        return []

    top = family_df.head(top_n_families)
    motifs = top["RecommendedAnchorMotif"].tolist()

    # preserve order, remove duplicates
    seen = set()
    out = []
    for m in motifs:
        if m not in seen:
            out.append(m)
            seen.add(m)
    return out


def discover_and_collapse_rescue_motifs(
    file_path,
    k=6,
    right_anchor_max_from_end=10,
    search_dist_range=(8, 25),
    anchor_center_dist=16,
    top_n_families=10,
):
    """
    End-to-end helper:
      1) scan motif windows
      2) summarize motif+distance
      3) collapse sliding-window families
      4) recommend anchor motifs
    """
    window_df = discover_rescue_motif_windows(
        file_path=file_path,
        k=k,
        right_anchor_max_from_end=right_anchor_max_from_end,
        search_dist_range=search_dist_range,
    )

    motif_summary_df = summarize_rescue_motif_windows(window_df)

    family_df, detailed_df = collapse_rescue_motif_families(
        window_df,
        anchor_center_dist=anchor_center_dist,
        preferred_k=k,
    )

    recommended = recommend_default_rescue_motifs(
        family_df,
        top_n_families=top_n_families,
    )

    return {
        "window_df": window_df,
        "motif_summary_df": motif_summary_df,
        "family_df": family_df,
        "detailed_df": detailed_df,
        "recommended_motifs": recommended,
    }
    
# =========================================================
# OPTIONAL: DISCOVER MOTIFS FROM YOUR GENERATED SET
# =========================================================
def discover_rescue_motifs(
    file_path,
    top_n=20,
    k=6,
    right_anchor_max_from_end=10,
    search_dist_range=(13, 18),
):
    """
    Scan sequences for 6-mers occurring 13-18 aa upstream of the terminal W anchor.
    This is useful for checking whether VTITAS-like motifs are really enriched
    in the current generated batch.
    """
    counter = Counter()
    n_total = 0
    n_with_right_w = 0

    for record in SeqIO.parse(file_path, "fasta"):
        seq = str(record.seq).upper()
        n_total += 1

        w_positions = [m.start() for m in re.finditer("W", seq)]
        terminal_ws = [p for p in w_positions if (len(seq) - 1 - p) <= right_anchor_max_from_end]
        if not terminal_ws:
            continue

        n_with_right_w += 1
        right_w = max(terminal_ws)

        for dist in range(search_dist_range[0], search_dist_range[1] + 1):
            start = right_w - dist
            if start < 0 or start + k > len(seq):
                continue
            counter[seq[start:start + k]] += 1

    motif_df = pd.DataFrame(counter.items(), columns=["Motif", "Count"])
    motif_df = motif_df.sort_values(["Count", "Motif"], ascending=[False, True]).reset_index(drop=True)
    motif_df["PercentOfAllSeqs"] = motif_df["Count"] / max(n_total, 1) * 100
    motif_df["PercentOfSeqsWithTerminalW"] = motif_df["Count"] / max(n_with_right_w, 1) * 100
    return motif_df.head(top_n)


# =========================================================
# CORE HELPERS
# =========================================================
def _hamming_similarity(a, b):
    if len(a) != len(b):
        return 0.0
    return sum(x == y for x, y in zip(a, b)) / len(a)


def _find_right_anchor_w(seq, max_from_end=10):
    """
    Use the rightmost W near the C-terminus as the FR4-entry-like anchor.
    """
    w_positions = [m.start() for m in re.finditer("W", seq)]
    terminal_ws = [p for p in w_positions if (len(seq) - 1 - p) <= max_from_end]
    return max(terminal_ws) if terminal_ws else None


def _collect_left_anchor_candidates(
    seq,
    right_w,
    motif_library,
    dist_range=(13, 18),
    min_similarity=1.0,
):
    """
    Search for VTITAS-like rescue motifs upstream of terminal W.
    We score candidate motifs by similarity and preferred spacing to W.
    """
    candidates = []

    for motif in motif_library:
        motif = motif.upper()
        k = len(motif)

        for dist in range(dist_range[0], dist_range[1] + 1):
            start = right_w - dist
            if start < 0 or start + k > len(seq):
                continue

            window = seq[start:start + k]
            sim = _hamming_similarity(window, motif)

            if sim >= min_similarity:
                # Mild preference for distances near 16 aa upstream of W
                score = sim - 0.02 * abs(dist - 16)
                candidates.append({
                    "Start": start,
                    "End": start + k,
                    "Window": window,
                    "MatchedMotif": motif,
                    "DistanceToW": dist,
                    "Similarity": sim,
                    "Score": score,
                    "MotifLength": k,
                })

    return candidates


def _resolve_left_anchor(candidates, ambiguity_delta=0.03):
    """
    Collapse motif hits by start position. If multiple distinct starts have
    near-identical top scores, mark as ambiguous.
    """
    if not candidates:
        return None, "no_left_anchor"

    best_by_start = {}
    for c in candidates:
        start = c["Start"]
        if start not in best_by_start or c["Score"] > best_by_start[start]["Score"]:
            best_by_start[start] = c

    collapsed = sorted(best_by_start.values(), key=lambda x: (x["Score"], x["Start"]), reverse=True)
    best = collapsed[0]

    near_ties = [c for c in collapsed[1:] if abs(c["Score"] - best["Score"]) <= ambiguity_delta]
    if near_ties:
        return None, "ambiguous_left_anchor"

    return best, "ok"


# =========================================================
# MAIN EXTRACTION FUNCTION
# =========================================================
def extract_cdr3_tiered_meta(
    seq,
    seq_id=None,
    canonical_map=None,
    motif_library=None,
    right_anchor_max_from_end=10,
    dist_range=(13, 18),
    min_similarity=DEFAULT_MIN_SIMILARITY,
    min_loop_len=5,
    max_loop_len=35,
    allow_rescue_fallback=True,
):
    """
    Two-tier logic:
      1) If canonical_map has an entry, use that.
      2) Otherwise rescue a graftable H3-like interval using:
         - terminal W as hard right anchor
         - VTITAS-like motif as FR3-proximal rescue motif

    Returns a metadata dict, not just the loop sequence.
    """
    seq = str(seq).upper()
    motif_library = motif_library or DEFAULT_RESCUE_MOTIFS

    meta = {
        "SequenceID": seq_id,
        "FullSequence": seq,
        "Method": None,
        "Status": None,
        "Sequence": None,          # extracted loop / H3-like interval
        "LoopLength": None,
        "RightAnchorW": None,
        "LeftAnchorStart": None,
        "LeftAnchorEnd": None,
        "LeftAnchorWindow": None,
        "MatchedMotif": None,
        "DistanceToW": None,
        "Similarity": None,
    }

    # -----------------------------
    # Tier 1: canonical (optional)
    # -----------------------------
    canonical_loop = None
    if canonical_map is not None and seq_id in canonical_map:
        canonical_loop = canonical_map[seq_id]

    if canonical_loop:
        canonical_loop = str(canonical_loop).upper()
        if (
            min_loop_len <= len(canonical_loop) <= max_loop_len
            and re.fullmatch(r"[ACDEFGHIKLMNPQRSTVWY]+", canonical_loop)
        ):
            meta.update({
                "Method": "canonical",
                "Status": "ok",
                "Sequence": canonical_loop,
                "LoopLength": len(canonical_loop),
            })
            return meta

        # If a canonical map is provided and rescue fallback is disabled,
    # do NOT attempt motif rescue for this sequence.
    if canonical_map is not None and not allow_rescue_fallback:
        meta.update({
            "Method": "canonical",
            "Status": "no_canonical_loop",
        })
        return meta
    # -----------------------------
    # Tier 2: motif rescue
    # -----------------------------
    right_w = _find_right_anchor_w(seq, max_from_end=right_anchor_max_from_end)
    if right_w is None:
        meta.update({
            "Method": "rescued",
            "Status": "no_right_anchor",
        })
        return meta

    meta["RightAnchorW"] = right_w

    candidates = _collect_left_anchor_candidates(
        seq=seq,
        right_w=right_w,
        motif_library=motif_library,
        dist_range=dist_range,
        min_similarity=min_similarity,
    )

    best, status = _resolve_left_anchor(candidates)
    if status != "ok":
        meta.update({
            "Method": "rescued",
            "Status": status,
        })
        return meta

    # Define the rescued H3-like interval as motif_end : right_w
    loop_start = best["End"]
    loop_end = right_w
    loop = seq[loop_start:loop_end]

    meta.update({
        "Method": "rescued",
        "LeftAnchorStart": best["Start"],
        "LeftAnchorEnd": best["End"],
        "LeftAnchorWindow": best["Window"],
        "MatchedMotif": best["MatchedMotif"],
        "DistanceToW": best["DistanceToW"],
        "Similarity": best["Similarity"],
    })

    if not re.fullmatch(r"[ACDEFGHIKLMNPQRSTVWY]+", loop):
        meta.update({
            "Status": "noncanonical_residue",
        })
        return meta

    if not (min_loop_len <= len(loop) <= max_loop_len):
        meta.update({
            "Status": "length_out_of_range",
            "LoopLength": len(loop),
        })
        return meta

    meta.update({
        "Status": "ok",
        "Sequence": loop,
        "LoopLength": len(loop),
    })
    return meta


def extract_cdr3_tiered(
    seq,
    seq_id=None,
    canonical_map=None,
    motif_library=None,
    **kwargs,
):
    """
    Backward-compatible wrapper:
    returns only the extracted loop string or None.
    """
    meta = extract_cdr3_tiered_meta(
        seq=seq,
        seq_id=seq_id,
        canonical_map=canonical_map,
        motif_library=motif_library,
        **kwargs,
    )
    return meta["Sequence"]


# =========================================================
# DATA LOADING
# =========================================================
def load_and_process(
    file_path,
    label,
    canonical_map=None,
    motif_library=None,
    return_audit=False,
    **extract_kwargs,
):
    """
    Compatible with your old workflow.
    Keeps the core columns:
      Sequence, Length, Hydrophobicity, Charge, Source
    and adds audit columns if available.
    """
    data = []
    audit_rows = []

    print(f"Processing {label}...")

    for record in SeqIO.parse(file_path, "fasta"):
        meta = extract_cdr3_tiered_meta(
            record.seq,
            seq_id=record.id,
            canonical_map=canonical_map,
            motif_library=motif_library,
            **extract_kwargs,
        )

        meta["Source"] = label
        audit_rows.append(meta.copy())

        loop = meta["Sequence"]
        if loop:
            try:
                analysis = ProteinAnalysis(loop)
                data.append({
                    "Sequence": loop,
                    "Length": len(loop),
                    "Hydrophobicity": analysis.gravy(),
                    "Charge": analysis.charge_at_pH(7.4),
                    "Source": label,

                    # extra audit-friendly metadata
                    "SequenceID": record.id,
                    "Method": meta["Method"],
                    "Status": meta["Status"],
                    "LeftAnchorWindow": meta["LeftAnchorWindow"],
                    "MatchedMotif": meta["MatchedMotif"],
                    "DistanceToW": meta["DistanceToW"],
                    "Similarity": meta["Similarity"],
                })
            except Exception:
                pass

    df = pd.DataFrame(data)
    audit_df = pd.DataFrame(audit_rows)

    if return_audit:
        return df, audit_df
    return df


# =========================================================
# AUDIT / FUNNEL SUMMARY
# =========================================================
def summarize_extraction(audit_df):
    total = len(audit_df)

    summary = [
        {"Stage": "Total input sequences", "Count": total},
        {"Stage": "Right terminal W found", "Count": audit_df["RightAnchorW"].notna().sum()},
        {"Stage": "Unique left-anchor rescue", "Count": audit_df["LeftAnchorWindow"].notna().sum()},
        {"Stage": "Recovered H3-like interval", "Count": (audit_df["Status"] == "ok").sum()},
        {"Stage": "Failed: no right anchor", "Count": (audit_df["Status"] == "no_right_anchor").sum()},
        {"Stage": "Failed: no left anchor", "Count": (audit_df["Status"] == "no_left_anchor").sum()},
        {"Stage": "Failed: ambiguous left anchor", "Count": (audit_df["Status"] == "ambiguous_left_anchor").sum()},
        {"Stage": "Failed: length out of range", "Count": (audit_df["Status"] == "length_out_of_range").sum()},
        {"Stage": "Failed: noncanonical residue", "Count": (audit_df["Status"] == "noncanonical_residue").sum()},
    ]

    out = pd.DataFrame(summary)
    out["Percent"] = out["Count"] / max(total, 1) * 100
    return out


def inspect_extraction_examples(audit_df, n=20):
    cols = [
        "SequenceID",
        "Status",
        "Method",
        "LeftAnchorWindow",
        "MatchedMotif",
        "DistanceToW",
        "Similarity",
        "Sequence",
    ]
    cols = [c for c in cols if c in audit_df.columns]
    return audit_df[cols].head(n)
```

```python
def seq_identity(a, b):
    if len(a) != len(b):
        return None
    return sum(x == y for x, y in zip(a, b)) / len(a)

seqs = df_gen["Sequence"].tolist()
nn_identities = []

for i, s1 in enumerate(seqs):
    best = -1
    for j, s2 in enumerate(seqs):
        if i == j:
            continue
        if len(s1) != len(s2):
            continue
        ident = seq_identity(s1, s2)
        if ident is not None and ident > best:
            best = ident
    if best >= 0:
        nn_identities.append(best)

nn_df = pd.DataFrame({"NearestNeighborIdentity": nn_identities})
nn_df.describe()
```

```python
import math
import pandas as pd
import matplotlib.pyplot as plt

def shannon_entropy(chars):
    probs = pd.Series(chars).value_counts(normalize=True)
    return -(probs * probs.map(math.log2)).sum()

def position_entropy_df(seq_list, source_label):
    """
    seq_list: list of equal-length sequences
    """
    if len(seq_list) == 0:
        return pd.DataFrame(columns=["Position", "Entropy_bits", "TopResidue", "TopResidueFreq", "Source"])

    L = len(seq_list[0])
    rows = []

    for pos in range(L):
        residues = [s[pos] for s in seq_list]
        vc = pd.Series(residues).value_counts(normalize=True)
        rows.append({
            "Position": pos + 1,
            "Entropy_bits": shannon_entropy(residues),
            "TopResidue": vc.idxmax(),
            "TopResidueFreq": vc.max(),
            "Source": source_label
        })

    return pd.DataFrame(rows)
```

```python
# len=10 only
nat10_seqs = df_train_len10["Sequence"].tolist()
gen10_seqs = df_gen_len10["Sequence"].tolist()

entropy_nat10 = position_entropy_df(nat10_seqs, "Natural len=10")
entropy_gen10 = position_entropy_df(gen10_seqs, "Generated len=10")

entropy_compare = pd.concat([entropy_nat10, entropy_gen10], ignore_index=True)
entropy_compare
```

```python
def seq_identity(a, b):
    if len(a) != len(b):
        return None
    return sum(x == y for x, y in zip(a, b)) / len(a)

def nearest_identity_to_reference(query_seqs, ref_seqs):
    rows = []
    for q in query_seqs:
        best = -1
        best_ref = None
        for r in ref_seqs:
            if len(q) != len(r):
                continue
            ident = seq_identity(q, r)
            if ident is not None and ident > best:
                best = ident
                best_ref = r
        rows.append({
            "Sequence": q,
            "NearestRefIdentity": best,
            "NearestRefSequence": best_ref
        })
    return pd.DataFrame(rows)
```

```python
gen_to_nat_raw = nearest_identity_to_reference(
    df_gen_len10["Sequence"].tolist(),
    df_train_len10["Sequence"].tolist()
)

print(gen_to_nat_raw["NearestRefIdentity"].describe())
print("Exact matches:", (gen_to_nat_raw["NearestRefIdentity"] == 1.0).sum())
print(">=0.9 identity:", (gen_to_nat_raw["NearestRefIdentity"] >= 0.9).sum())
print(">=0.8 identity:", (gen_to_nat_raw["NearestRefIdentity"] >= 0.8).sum())
```

```python
import pandas as pd
import numpy as np

def concentration_metrics(seq_series, label):
    """
    seq_series: pandas Series of sequences
    """
    counts = seq_series.value_counts()
    probs = counts / counts.sum()

    hhi = (probs ** 2).sum()
    effective_n = 1 / hhi if hhi > 0 else np.nan

    def topk_share(k):
        return probs.iloc[:k].sum() * 100

    out = pd.DataFrame({
        "Group": [label],
        "Total": [len(seq_series)],
        "Unique": [seq_series.nunique()],
        "UniquePercent": [seq_series.nunique() / len(seq_series) * 100],
        "Top1Percent": [topk_share(1)],
        "Top5Percent": [topk_share(5)],
        "Top10Percent": [topk_share(10)],
        "Top20Percent": [topk_share(20)],
        "HHI": [hhi],
        "EffectiveN": [effective_n],
    })
    return out


def top_sequence_table(seq_series, label, top_n=20):
    counts = seq_series.value_counts().reset_index()
    counts.columns = ["Sequence", "Count"]
    counts["Percent"] = counts["Count"] / counts["Count"].sum() * 100
    counts["CumulativePercent"] = counts["Percent"].cumsum()
    counts["Group"] = label
    return counts.head(top_n)
```

```python
conc_nat10 = concentration_metrics(df_train_len10["Sequence"], "Natural len=10")
conc_gen10 = concentration_metrics(df_gen_len10["Sequence"], "Generated len=10")

conc_compare = pd.concat([conc_nat10, conc_gen10], ignore_index=True)
conc_compare
```

```python
# 2. Pre-process for Logo (Align to Center)
# Logos need fixed length. We will align by the 'C' anchor and pad the end.
def pad_sequences(sequences, length=15):
    padded = []
    for s in sequences:
        if len(s) > length: s = s[:length] # Truncate long ones
        else: s = s + '-' * (length - len(s)) # Pad short ones
        padded.append(s)
    return padded

max_len = 18
train_padded = pad_sequences(train_loops, max_len)
gen_padded = pad_sequences(gen_loops, max_len)

# 3. Create Logos
fig, axes = plt.subplots(2, 1, figsize=(10, 6))

# Train Logo
train_counts = logomaker.alignment_to_matrix(train_padded)
logo_train = logomaker.Logo(train_counts, ax=axes[0], color_scheme='chemistry')
axes[0].set_title("Training Data: CDRH3 Sequence Grammar", fontsize=14)
axes[0].set_ylabel("Bits")

# Gen Logo
gen_counts = logomaker.alignment_to_matrix(gen_padded)
logo_gen = logomaker.Logo(gen_counts, ax=axes[1], color_scheme='chemistry')
axes[1].set_title("GAN Generated: CDRH3 Sequence Grammar", fontsize=14)
axes[1].set_ylabel("Bits")

plt.tight_layout()
plt.savefig(OUTPUT_PATH + "Figure_B_SequenceLogos_new.png", dpi=300)
plt.show()
```
